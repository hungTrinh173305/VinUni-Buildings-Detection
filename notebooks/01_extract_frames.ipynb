```python
import os
import glob
import cv2

RAW_DIR        = "/content/drive/MyDrive/COMP4010_finalprj/building_K/building_k_raw"
OUTPUT_DIR     = os.path.join(RAW_DIR, "frames_new")
FPS            = 1
BLUR_THRESHOLD = 100

os.makedirs(OUTPUT_DIR, exist_ok=True)

def is_blurry(frame, threshold):
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    return cv2.Laplacian(gray, cv2.CV_64F).var() < threshold

videos  = sorted(glob.glob(os.path.join(RAW_DIR, "new*.mp4")))
summary = []

for video_path in videos:
    name = os.path.splitext(os.path.basename(video_path))[0]
    cap  = cv2.VideoCapture(video_path)
    fps  = cap.get(cv2.CAP_PROP_FPS)
    interval = max(1, int(round(fps / FPS)))

    frame_count = saved = skipped = 0

    while True:
        ok, frame = cap.read()
        if not ok:
            break
        if frame_count % interval == 0:
            if is_blurry(frame, BLUR_THRESHOLD):
                skipped += 1
            else:
                saved += 1
                cv2.imwrite(os.path.join(OUTPUT_DIR, f"{name}_frame_{saved:04d}.jpg"), frame)
        frame_count += 1

    cap.release()
    summary.append((name, saved, skipped))
    print(f"{name}: saved {saved}, skipped {skipped} blurry")

print(f"\nTotal saved : {sum(s for _, s, _ in summary)}")
print(f"Total skipped: {sum(sk for _, _, sk in summary)}")
print(f"Output: {OUTPUT_DIR}")
```
